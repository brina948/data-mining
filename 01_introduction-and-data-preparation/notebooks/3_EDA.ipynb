```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Buat data normal
np.random.seed(42)
data_normal = np.random.normal(loc=50, scale=5, size=100)

# Tambahkan beberapa outlier
outliers = np.array([80, 85, 90])
data_with_outliers = np.concatenate([data_normal, outliers])

# Simpan ke DataFrame
df = pd.DataFrame({'Value': data_with_outliers})
```

```python
plt.figure(figsize=(7,4))
plt.hist(df['Value'], bins=15, edgecolor='black')
plt.title('Histogram – Deteksi Outlier')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.boxplot(df['Value'], vert=False)
plt.title('Boxplot – Deteksi Outlier')
plt.xlabel('Value')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1️⃣ Buat data dengan outlier
np.random.seed(42)
data = np.random.normal(100, 10, 100)
data = np.append(data, [180, 190, 200])  # tambahkan outlier

data = pd.DataFrame({'Price': data})
```

```python
data.head()
```

```python
# 4️⃣ Visualisasi perbandingan
plt.figure(figsize=(10,4))

plt.subplot(1,3,1)
plt.boxplot(data['Price'], vert=True, patch_artist=True)
plt.title('Before Handling')
```

```python
# Tentukan batas
lower_limit = data['Price'].quantile(0.05)
upper_limit = data['Price'].quantile(0.95)
```

```python
# Trimming
data_trimmed = data[(data['Price'] <= upper_limit) & (data['Price'] >= lower_limit)]
```

```python
plt.subplot(1,3,2)
plt.boxplot(data_trimmed, vert=True, patch_artist=True)
plt.title('After Trimming')
```

```python
# Tentukan batas persentil
lower_limit = data['Price'].quantile(0.05)
upper_limit = data['Price'].quantile(0.95)
```

```python
# Winsorizing
data_Winsor = np.where(data['Price'] > upper_limit, upper_limit,
                              np.where(data['Price'] < lower_limit, lower_limit, data['Price']))
```

```python
# Visualisasi hasil boxplot
plt.figure(figsize=(6,3))
plt.boxplot(data_Winsor, vert=True, patch_artist=True)
plt.title('After Winsorizing')
plt.show()
```
